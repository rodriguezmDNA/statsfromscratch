### Flippin' coins

From Bioinformatics Algorithms: An Active Learning Approach book, chapter 8.

Flipping a coing we get "H" 157 consecutive times. Is the coin biased?

If we flip a `biased` coin `n` times, what is the probability $\theta$ that a signle flip will be "H"?




<bloquote>
<em>Stop and think:</em>

For each value of $\theta$ between 0 and 1, you can compute the probability of a given sequence of flips. <br>
How would you estimate the value of $\theta$ maximizing this probability after watching a series of flips where the coin lands on "H" `i` out of `n` times?
</bloquote>

For a given series of flips, $ \theta = \frac{i}{n} $ maximizes the probability of observing that series of flips.

This likelihood can be more formally obtained from the binomial distribution:

$$ p( \text{observing \textit{i} Heads}) = \binom{n}{i} \theta^i (1-\theta)^{n-i} $$

* $\binom{n}{i}$ is the binomial coefficient, which is the number of ways to choose `i` items from `n` items.
* $\theta^i$ is the probability of observing `i` Heads.
* $(1-\theta)^{n-i}$ is the probability of observing `n-i` Tails.

The log likelihood is easier to work with, so we can take the log of the likelihood with respect to $\theta$:

$$  L(\theta) = \log \binom{n}{i} + i \log \theta + (n-i) \log (1-\theta) $$

* $\log \binom{n}{i}$ is the log of the binomial coefficient and it is contant with respect to $\theta$. We can ignore it when maximizing the likelihood.

Since we are maximizing the likelihood, ignore the constant term and derive the rest of the equation. 


After deriving the log likelihood, we can set it to zero to find the maximum likelihood estimate of $\theta$:

$$ \frac{dL}{d\theta} = \frac{i}{\theta} - \frac{n-i}{1-\theta} = 0 $$
$$ \frac{i}{\theta} = \frac{n-i}{1-\theta} $$
$$ i - i\theta = n\theta - i\theta $$
$$ i = n\theta $$
$$ \theta = \frac{i}{n} $$


$\theta = \frac{i}{n}$ is the maximum likelihood estimate of $\theta$ given a series of `n` flips where `i` of them are "H".

What happens if we have more than one coin? They might look identical but their underlying "biases" are unkown to us. 

All we have is the observed series of flips. How can we estimate the biases of each coin?


$\theta_{A}$ and $\theta_{B}$ are the biases of coin A and coin B, respectively. Also called as _parameters_.

To make it simpler, every `n` flips we can either keep the coin or switch to the other coin.

After five sequences of `n=10` flips we have the following observations:


$$Data = (Data1, Data2, Data3, Data4, Data5) = (0.4, 0.9, 0.8, 0.3, 0.7)$$

If we knew that $coin_{A}$ was used for the 1st and 4th sets, then $\theta_{A} = \frac{0.4+0.3}{2}=0.35$. <br>
Knowing this, we can estimate $\theta_{B}$ as $\frac{0.9+0.8+0.7}{3}=0.8$.

In this scenario the hidden vector of coin assignments is $Z = (A, B, B, A, B)$.

To estimate the bias of A $\theta_{A}$ and the bias of B $\theta_{B}$, we can multiply each observed data point by the corresponding coin assignment as a weighted average:

$$\theta_{A} = \frac{\sum_{i=1}^n h_i \times d_i}{\sum_{i=1}^n h_i}$$

$h_{i}$ of $A$ would be $ (1, 0, 0, 1, 0)$ and $d_{i}$ would be $(0.4, 0.9, 0.8, 0.3, 0.7)$.

Thus $$\theta_{A} = \frac{ (1\cdot0.4) + (0\cdot0.9) + (0\cdot0.8) + (1\cdot0.3) + (0\cdot0.7)}{1+0+0+1+0} = 0.35$$

Similarly, $h_{i}$ of $B$ would be $ (0, 1, 1, 0, 1)$ and $d_{i}$ would be $(0.4, 0.9, 0.8, 0.3, 0.7)$:

 $$\theta_{B} = \frac{ (0\cdot0.4) + (1\cdot0.9) + (1\cdot0.8) + (0\cdot0.3) + (1\cdot0.7)}{0+1+1+0+1} = 0.8$$


Given this information we can determine that the parameters are: $$\theta_{A} = 0.35 \text{ , } \theta_{B} = 0.8$$

If we know the parameters, we can determine which coin was more likely to be used for a given sequence of flips.

For example given the parameters $\theta_{A} = 0.6$ and $\theta_{B} = 0.82$ and the sequence $THHHTHHHTH$ we can calculate the probabilities of the sequence given each coin:

Probability of the hidden sequence given coin A (the bias is towards $H$): 

3 Tails and 7 Heads: $$P(THHHTHHHTH | A) = \theta_{A}^7 \times (1-\theta_{A})^ 3 = 0.6^7 \times 0.4^3 = 0.00179$$

Probability of the hidden sequence given coin B (the bias is towards $H$):

3 Tails and 7 Heads: $$P(THHHTHHHTH | B) = \theta_{B}^7 \times (1-\theta_{B})^ 3 = 0.82^7 \times 0.18^3 = 0.00145$$

Thus, the sequence is more likely to have been generated by coin A.

We can generalize the above example to a larger number of coins and sequences of flips.

The probability of the data (the hidden vector) given the parameter (coin bias) $Pr(Data_{i}|\theta)$ is the **condiitonal probability** of the data given the parameter.

$$Pr(Data_{i}|\theta) = \theta^{i} \times (1-\theta)^{n-i}$$

Where $i$ is the number of $H$ in the sequence and $n$ is the number of flips.


If the $HiddenVector$ is known but the parameters are not, we can reverse engineer the most likely $Parameters$ given the data.

$$(Data, HiddenVector, ?) \rightarrow Parameters$$

If $Parameters$ is known and $HiddenVector$ is not, we can determine the most likely $HiddenVector$ given the data.

$$(Data, ?, Parameters) \rightarrow HiddenVector$$

However the initial problem is that both $HiddenVector$ and $Parameters$ are unknown. We can solve this problem by using the Expectation-Maximization (EM) algorithm.

$$ (Data, ?, ?) \rightarrow ???$$

### Conditional probability

THe conditional probability of generating $Darta$ given $HiddenVector$ and $Parameters$ is:


$$Pr(Data, HiddenVector | Parameters) = \prod_{i=1}^{n} Pr(Data_{i} | HiddenVector, Parameters )$$

The computational problem we want to solve is find $HiddenVector$ and $Parameters$ that maximize $Pr(Data, HiddenVector | Parameters)$.

If we start from a random arbitrary set of $Parameters$ we can immediately reconstruct the most likely $HiidenVector$:

$$(Data, ?, Parameters) \rightarrow HiddenVector$$

As soon as we know $Hiddenvector$ we will question the validity of the initial choice of $Parameters$ and update them accordingly:

$$ (Data, HiddenVector, ?) \rightarrow Parameters$$


## Lloyds algorithm

```python
### Choose the hidden vector that is most likely to have generated the data

pA = 0.6
pB = 0.82

data = [0.4, 0.9, 0.8, 0.3, 0.7]

values_at_epoch = [(pA, pB)]

for _ in range(10):

    hidden_vector = [
        1 if abs(x - pA) <= abs(x - pB) else 0
        for x in data
    ]

    pA = (
        sum(x for (x, h) in zip(data, hidden_vector) if h == 1) 
        / 
        sum(hidden_vector)
    )

    pB = (
        sum(x for (x, h) in zip(data, hidden_vector) if h == 0) 
        / 
        (len(hidden_vector) - sum(hidden_vector))
    )

    values_at_epoch.append((pA, pB))

    ## Apply early stopping if convergence is reached
    if len(values_at_epoch) > 1 and values_at_epoch[-1] == values_at_epoch[-2]:
        break

values_at_epoch
```

### Expectation-Maximization (EM) algorithm

The EM algorithm is a generalization of the Lloyds algorithm for clustering. It is used to find the most likely parameters of a model given the data.

Given $Parameters = (\theta_{A}, \theta_{B})$ we can hard-code the $HiddenVector$ by comparing $Pr(Data_{i} | \theta_{A})$ and $Pr(Data_{i} | \theta_{B})$. 
The problem is that we are not certain which coin was used for each sequence of flips.

If $Pr(Data_{i} | \theta_{A})$ is approximately equal to $Pr(Data_{i} | \theta_{B})$ then our confidence that coin B was used would be approximately 50%.
If it were much larger, then our confidence that coin B was used would be much higher.

We can refer to this 'confidence' as the 'responsibility' of a coin for a given sequence of flips.

The responsibility of coin A and coin B should sum to 1.


In terms of clustering, if a data point is a midpoint between two centers, then the responsibility of each center is 0.5.

Given $Parameters = (\theta_{A}, \theta_{B}) = (0.6 , 0.82)$ and the sequence $(T, H, H, H, T, H, H, H, T, H)$, we can calculate the responsibilities of each coin for each sequence of flips.


There are 7 $H$ and 3 $T$ in the sequence.

The probability of this outcome given coin A is: $Pr(Data_{i} | \theta_{A}) = 0.6^7 \times 0.4^3 = 0.00179$, <br>
and the probability of this outcome given coin B is: $Pr(Data_{i} | \theta_{B}) = 0.82^7 \times 0.18^3 = 0.00145$.

We *__rigidly__* assign the coin with the highest probability to each sequence of flips.

Since coin A is more likely to have generated the sequence, we should assign a higher responsibility to coin A than to coin B. <br>
One way to do this is to calculate the relative probabilities of each coin for each sequence of flips:

$$Responsibility_{A} = \frac{Pr(Data_{i} | \theta_{A})}{Pr(Data_{i} | \theta_{A}) + Pr(Data_{i} | \theta_{B})} = \frac{0.00179}{0.00179 + 0.00145} = 0.55$$

and,

$$Responsibility_{B} = \frac{Pr(Data_{i} | \theta_{B})}{Pr(Data_{i} | \theta_{A}) + Pr(Data_{i} | \theta_{B})} = \frac{0.00145}{0.00179 + 0.00145} = 0.45$$


----


Instead of a $HiddenVector$ we have a $2x5$ matrix of responsibilities that can be constructed from $Data$ and $Parameters$.

This is the E-step of the EM algorithm.

For a set of data points $Data = (Data_{1}, Data_{2}, ... , Data_{n})$ and a set of k coins with biases $Parameters = (\theta_{A}, ... , \theta_{k})$ we can calculate the responsibilities of each coin for each sequence of flips.


The $k\times n$ $HiddenMatrix$ is defined as:

$$HiddenMatrix_{i,j} = \frac{Pr(Data_{i} | \theta_{i})}{\sum_{1 \leq t \leq k} Pr(Data_{i} | \theta_{t})}$$

```python
n_flips = 10

Data = [0.4, 0.9, 0.8, 0.3, 0.7]
thetaA = 0.6
thetaB = 0.82

parameters = [thetaA,thetaB]

# Compute the hidden matrix

def p_data(theta, n, n_flips):
    return theta**n * (1-theta)**(n_flips-n)

hidden_matrix = np.zeros((len(Data),2))
for i, x in enumerate(Data):
    # Interpret e.g. 0.4 as 4 heads out of 10
    k = int(round(x * n_flips))
    # Denominator: sum of likelihoods over both parameters
    denom = sum(p_data(th, k, n_flips) for th in parameters)
    for j in range(len(parameters)):
        num = p_data(parameters[j], k, n_flips) # This is the likelihood of the data given the parameter
        hidden_matrix[i, j] = np.round(num / denom,2) # This is the posterior probability or responsibility of each parameter

hidden_matrix = hidden_matrix.T
hidden_matrix
```

```python
## Maximization step: recompute the parameters

parameters = [ np.sum([np.prod(i) for i in zip(row, Data)])/np.sum(row) for row in hidden_matrix]
parameters = [np.round(p,3) for p in parameters]
parameters
```
